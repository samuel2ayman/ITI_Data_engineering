```python
def greatest_number(x,y,z):
    numbers=[x,y,z]
    result=numbers[0]
    for i in range(2):
        if result >=numbers[i+1]:
            continue
        else :
            result=numbers[i+1]
    return(result)
print(greatest_number(9,35,7))
```

```python
def sum_series(n):
    if n % 2==0:
        return -n/2
    elif n % 2 != 0:
        return (n+1)/2

print(sum_series(2))
```

```python
def factorial(n):
    fact=1
    for i in range(n,0,-1):
        fact*=i
    print(f'!{n} = ', end='')
    return fact
    
    
print(factorial(5))
        
```

```python
def palindrome_check(word):
    reversed_word=word[::-1]
    if reversed_word== word:
        print(f'{word} is polindrome')
    else:
        print(f'{word} is not polindrome')
palindrome_check('pad')
```
