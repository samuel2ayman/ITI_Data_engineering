```python
# - Calculate the average grade for each student.
students = {
    "Alice": [85, 90, 78],
    "Bob": [92, 88, 79],
    "Charlie": [76, 81, 85],
    "Diana": [95, 89, 94],
    "Ethan": [70, 75, 80]
}

def avg_grades(students):
    averages={}
    for name,grades in students.items():
        avg=sum(grades)/len(grades)
        averages[name]=avg
    return averages
avg_dict=avg_grades(students)
for name,avg_grade in avg_dict.items():
        print(f"{name} average grade is: {avg_grade}")
```

```python
# - Calculate the highest average grade 
def highest_avg(avg_dict):
    avg_list=[avg for avg in avg_dict.values()]
    max_avg=max(avg_list)
    for name,avg_grade in avg_dict.items():
        if avg_grade==max_avg:
            print(f"{name} has max average = {avg_grade}")

highest_avg(avg_dict)
    
```

```python
#2. **Shopping Cart:**

shopping_cart = {
    "Laptop": 899.99,
    "Headphones": 49.99,
    "Mouse": 19.99,
    "Monitor": 159.99,
    "Keyboard": 29.50,
    "USB Cable": 8.99,
    "Desk Lamp": 24.99
}
#Add items to the cart.
def add_element(element,price):
    shopping_cart[element]=price
    return shopping_cart
    
print(add_element('cable',55.5))
 
# - Calculate the total price of items in the cart.
def total_price(shopping_cart):
    values=list(shopping_cart.values())
    sum_values=sum(values)
    return sum_values
    
total=total_price(shopping_cart)
print(f"Total price of items in cart : {total}")
```

```python
#2. **Inventory Management:**
# - Update stock after a sale.
# Inventory system with dummy data

inventory = {
    "Laptop": {
        "price": 15000,
        "quantity": 5
    },
    "Mouse": {
        "price": 250,
        "quantity": 40
    },
    "Keyboard": {
        "price": 450,
        "quantity": 25
    },
    "Monitor": {
        "price": 3200,
        "quantity": 10
    },
    "USB Cable": {
        "price": 120,
        "quantity": 60
    }
}
```

```python
# - Update stock after a sale.
def update_after_sale(product_sal,quantity,inventory):
    for product,info in inventory.items():
        if product==product_sal:
            info['quantity']-=quantity
    return inventory

print(inventory)
update_after_sale('Mouse',5,inventory)      
print(inventory)
```

```python
# - Restock items.
def update_after_restock(product_res,quantity,inventory):
    for product,info in inventory.items():
        if product==product_res:
            info['quantity']+=quantity
    return inventory

print(inventory)
update_after_restock('Mouse',5,inventory)      
print(inventory)
```

```python
#5. **Two-Sum Problem: first two numbers which their sum give required number
int_list=[1,3,4,5,7,9,2,6]
def two_sum(int_list,num):
    numbers=[]
    for i in int_list:
        for j in int_list:
            if i+j==num and int_list.index(i)!=int_list.index(j):
                numbers.append(i)
                numbers.append(j)
        if len(numbers)>0:
            break
    return numbers

two_sum(int_list,11)
                
    
```

```python
#5. **Two-Sum Problem: All two numbers which their sum give required number
int_list=[1,3,4,5,7,9,2,6]
def two_sum(int_list,num):
    numbers=[]
    for i in int_list:
        if i in numbers:
            continue
        else:
            for j in int_list:
                if i+j==num :
                    if j in numbers:
                       continue
                    else:
                        numbers.append(i)
                        numbers.append(j)
                
        
    return numbers

two_sum(int_list,11)
                
    
```
