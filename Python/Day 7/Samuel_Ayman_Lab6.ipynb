```python
#Problem 1: File Reader with Exception Handling

from pathlib import Path
def file_reader():
    
    while True:
        file_name=input("Enter file name: ")
        path=Path(file_name)
        try:
            contents=path.read_text()
        except FileNotFoundError:
            print(f"Error: File {file_name} not found.")
            try_again=input("Would you like to try again? (yes/no): ").lower()
            if try_again=='yes':
                continue
            else:
                break
        else:
            print(f"contents of {file_name}")
            print(contents)
            break
    
file_reader()
    
```

```python
#Problem 2: Write Data to a File
def write_to_file():
    file_name=input("Enter file name: ")
    text=input("Enter text to save: ")
    path=Path(file_name)        
    try:
        path.write_text(text)
    except PermissionError:
        print(f"Error: Permission denied to write to {path}.")
    else:
        print(f"Data saved successfully to {path}")
        print(f"Contents of {path}:")
        contents=path.read_text()
        print(contents)
        
write_to_file()
```

```python
#problem 3 Store Data in JSON Forma

import json
data = {
"name": "Alice",
"age": 25,
"city": "New York" }

file_name='data.json'
path=Path(file_name)

def write_to_json(path,data):
    try:
        contents=json.dumps(data)
        path.write_text(contents)
    except IOError:
        print(f"Error writing to file {path}")
    else:
        print(f"Data saved successfully to {path}")
        print(f"Contents of {path}:")
        cont=path.read_text()
        file_data=json.loads(cont)
        # print(type(cont))
        # print(type(file_data))
        print(file_data)
    
write_to_json(path,data)

```
