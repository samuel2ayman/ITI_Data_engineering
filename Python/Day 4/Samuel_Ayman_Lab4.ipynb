```python
def list_operations():
    """add into list and remove from list from 1 and add 1 in the next """
    integers=[]
    count=1
    while True:
        user_input=input('')
        if user_input =='d' or user_input =='x' or user_input =='r':
            if user_input=='d':
                integers.append(count)
                count+=1
                print(integers)
            elif user_input=='r':
                if len(integers)==0:
                    print('List is empty')
                else:
                    integers.pop()
                    count-=1
                    print(integers)
            elif user_input=='x':
                break
    
        else:
            print('Not valid input please try again')
```

```python
list_operations()
```

```python
#Task 2
int_list=[3,3,2,2,6,6,8]
def distinct_numbers(int_list):
    my_list=[]
    for i in int_list:
        if i in my_list:
            continue
        else:
            my_list.append(i)
    sorted_list=sorted(my_list)
    return sorted_list
     
    
print(distinct_numbers(int_list))
```

```python
#Task 3
strings=['adele','mark','dorothy','elevength','qqqwwweee']
lengths=[len(string) for string in strings]

def length_of_longest(strings):
    """gives longest string in length"""
    max_length=max(lengths)
    return max_length
    
```

```python
max_lenth_index=lengths.index(length_of_longest(strings))
print(strings[max_lenth_index])
```

```python
#Task 4
result=[]
def all_the_longest(strings):
    """gives longest strings in length"""
    max_length=length_of_longest(strings)
    for i in strings:
        if len(i)==max_length:
            result.append(i)
    return result
            
            
```

```python
print(all_the_longest(strings))
```

```python
#Task 5
def same_word():
    """ask user to input word if user inputs one word twice program 
    ends and print different words"""
    list_of_words=[]
    while True:
        word=input('word : ')
        if word in list_of_words:
            print(f'you typed in {len(list_of_words)} different words')
            break
        else:
            list_of_words.append(word)
        
        
```
