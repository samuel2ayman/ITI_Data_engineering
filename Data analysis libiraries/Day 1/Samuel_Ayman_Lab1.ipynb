# 7. Write a NumPy program to combine the two 2-dimensional arrays using hstack and vstack.
# Array 1: [[0, 1, 3], [5, 7, 9]]
# Array 2: [[0, 2, 4], [6, 8, 10]]


```python
array_1=np.array([[0, 1, 3], [5, 7, 9]])
array_2=np.array([[0, 2, 4], [6, 8, 10]])
horizontal_merge=np.hstack((array_1,array_2))
print(horizontal_merge)
```

```python
horizontal_merge=np.hstack((array_1,array_2))
vertical_merge=np.vstack((array_1,array_2))
print(vertical_merge)
```

# 10.Write a NumPy program to reverse an array [using two different methods].

```python
#first method
my_arr=np.array([10,20,30,40])
reversed_arr=my_arr[::-1]
print(my_arr)
print(reversed_arr)
```

```python
#second method
my_arr=np.array([10,20,30,40])
reversed_list=[my_arr[i] for i in range(len(my_arr)-1,-1,-1)]
reversed_arr=np.array(reversed_list)
print(my_arr)
print(reversed_arr)
```
