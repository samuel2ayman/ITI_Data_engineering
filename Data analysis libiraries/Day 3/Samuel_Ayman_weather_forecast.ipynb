```python
import pandas as pd
pd.set_option('display.max_columns', None)
```

```python
#complex json file structure so we need to normalize first as read csv cannot read
import json
with open('weather_forecast.json') as f:
    data=json.load(f)
```

```python
weather1=pd.json_normalize(data,record_path='list',sep='_').drop(columns='weather')
weather_1
```

```python
weather_2=pd.json_normalize(data,record_path=['list','weather'],meta=[['list','dt']]).drop_duplicates(subset='list.dt').reset_index(drop=True)
weather_2
```

```python
weather=pd.concat([weather_1,weather_2],axis=1).drop(columns='list.dt')
weather
```
