## Exercise 1 — Patient Blood Sugar Readings (Series)

A Series of glucose readings (mg/dL) for 10 patients, indexed by patient name.

```python
import pandas as pd

blood_sugar = pd.Series(
        [82, 145, 200, 310, 95, 178, 260, 110, 340, 88],
        index=["Ali", "Sara", "Omar", "Mona", "Khaled",
               "Nour", "Tamer", "Hana", "Ramy", "Dina"],
        name="glucose_mg_dl"
)

print(blood_sugar)
```

### Task 1.1 — Label-Based Selection
Select **Omar's** glucose reading using his name as a label.

```python
blood_sugar['Omar']
```

### Task 1.2 — Positional Indexing
Select the **first 4 patients** using positional indexing.

```python
blood_sugar.iloc[0:4]
```

### Task 1.3 — Boolean Masking
Filter patients whose glucose is **above 200** (diabetic range).  
How many patients fall in this range?

```python
blood_sugar[blood_sugar> 200]
```

```python
(blood_sugar> 200).sum()
```

### Task 1.4 — Fancy Indexing
Select readings for **Sara, Nour, and Dina** using a list of labels in a single line.

```python
blood_sugar[['Sara','Nour','Dina']]
```

## Exercise 2 — Hospital Encounters (DataFrame)

A DataFrame of patient hospital encounters. The index is set to `encounter_id`.

| encounter_id | patient_name | age_group | time_in_hospital | num_medications | readmitted |
|---|---|---|---|---|---|
| 1001 | Ali Hassan | [30-40) | 3 | 12 | No |
| 1002 | Sara Nader | [60-70) | 7 | 21 | >30 |
| 1003 | Omar Fathi | [50-60) | 2 | 8 | No |
| 1004 | Mona Said | [70-80) | 11 | 30 | <30 |
| 1005 | Khaled Zaki | [40-50) | 5 | 15 | No |
| 1006 | Nour Adel | [60-70) | 9 | 25 | <30 |
| 1007 | Tamer Wael | [20-30) | 1 | 5 | No |

```python
data = {
    "encounter_id":    [1001, 1002, 1003, 1004, 1005, 1006, 1007],
    "patient_name":    ["Ali Hassan", "Sara Nader", "Omar Fathi",
                        "Mona Said", "Khaled Zaki", "Nour Adel", "Tamer Wael"],
    "age_group":       ["[30-40)", "[60-70)", "[50-60)", "[70-80)",
                        "[40-50)", "[60-70)", "[20-30)"],
    "time_in_hospital": [3, 7, 2, 11, 5, 9, 1],
    "num_medications":  [12, 21, 8, 30, 15, 25, 5],
    "readmitted":       ["No", ">30", "No", "<30", "No", "<30", "No"]
}

df = pd.DataFrame(data)
df = df.set_index("encounter_id")  # encounter_id becomes the index

df
```

### Task 2.1 — Column Selection
Select **only** the `patient_name` and `readmitted` columns for all rows.

```python
df[['patient_name','readmitted']]
```

### Task 2.2 — Row Selection with `.loc`
Using `.loc`, retrieve the complete row for encounter **1004** (Mona Said).

```python
df.loc[1004,:]
```

### Task 2.3 — Positional Slicing with `.iloc`
Using `.iloc`, retrieve the **last 3 rows** and only the **first 2 columns**.

```python
df.iloc[4:,:2]
```

### Task 2.4 — Combined Boolean Filtration
Filter patients who were **readmitted in under 30 days** AND spent **more than 5 days** in hospital.  
Show only their `patient_name` and `time_in_hospital`.

```python
df.loc[(df['readmitted'] == '<30') & (df['time_in_hospital'] > 5),['patient_name','time_in_hospital']]
```
